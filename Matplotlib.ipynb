It allows us to turn numerical and categorical data into visual forms such as:
1. Line charts
2. Bar charts
3. Scatter plot
4. Histograms
5. Box plots
6. Area charts

## Why do we use Matplot?
It helps us to:
1. Understand data more easily.
2. Identify trends.
3. Compare values.
4. Find patterns.
5. Understand distributions.
6. Communicate results visually.

# 1. Line plot: 
### A line plot is visualization used to show how a numerical value changes across an ordered sequence specially over time.

## Line chart can help us identify:
1. Trends - General direction of changes over time.
2. Patterns - Repeatable or noticeable movements.
3. Peaks and Troughs - High and low point in a sequence.
4. Rate of change - How quickly or slowly a values changes.

```python
# e.g.

months = ["January", "February", "March", "April", "May", "June"]
sales = [100, 150, 250, 85, 90, 130]

plt.plot(months, sales)
plt.show()
```

```python
# X-label, Y-label and Title:

months = ["January", "February", "March", "April", "May", "June"]
sales = [100, 150, 250, 85, 90, 130]

plt.plot(months, sales)
plt.xlabel("Year 2026")
plt.ylabel("sales")
plt.title("Monthly sales of 2026")
# plt.grid(axis="x") # If you want grid only in x-axis
# plt.grid(axis="y") # If you want grid only in y-axis
plt.grid() # If you want grid then use this.
plt.show()
```

```python
# Marker: There are several types of marker some of then are "o" and "*", etc.
# Color: It is used to change the line-color.

months = ["January", "February", "March", "April", "May", "June"]
sales = [100, 150, 250, 85, 90, 130]

plt.plot(months, sales, marker="o", color="blueviolet")
plt.show()
```

```python
# Multiline plot: It is used to plot multiple lines in a single graph.

months = ["January", "February", "March", "April", "May", "June"]
sales1 = [100, 150, 250, 85, 90, 130]
sales2 = [40, 45, 105, 202, 130, 100]

plt.plot(months, sales1, marker="o", color="blueviolet", label="smart-watch")
plt.plot(months, sales2, marker="o", color="goldenrod", label="phone")
plt.legend() # This is used to show label for each line for this to work "label" argument is mandatory in plots.
plt.show()
```

# 2. Bar charts: 
### A bar chart is visualization used to compare a numerical value across different categories.

## Bar chart can help us identify:
1. Comparison - Difference between categories.
2. Ranking - Which categories has higher or lower values.
3. Magnitude - The numerical size associated with each category.

```python
#e.g.

product = ["smart-watch", "mobile", "tv", "pencil", "laptop"]
price = [2000, 25000, 17000, 2000, 75000]

plt.bar(product, price)
plt.show()
```

```python
# Color, X-label, Y-label, Title, Grid:

product = ["smart-watch", "mobile", "tv", "pencil", "laptop"]
price = [2000, 25000, 17000, 2000, 75000]

plt.bar(product, price, color="purple")
plt.title("Electronics-Prices")
plt.grid()
plt.xlabel("products")
plt.ylabel("prices")
plt.show()
```

```python
# Changing edge color and showing values on top of bar.

product = ["smart-watch", "mobile", "tv", "pencil", "laptop"]
price = [2000, 25000, 17000, 2000, 75000]
bars = plt.bar(product, price, color="lightgreen", edgecolor="black") # We need to store plt.bar() in a variable to show values. There are alternate ways to do the same too.
plt.title("Electronics-Prices")
plt.xlabel("products")
plt.ylabel("prices")
plt.bar_label(bars)
plt.show()
```

```python
# Coloring each bar with different color.

product = ["smart-watch", "mobile", "tv", "pencil", "laptop"]
price = [2000, 25000, 17000, 2000, 75000]
colors = ["purple", "goldenrod", "green", "blue", "yellow"]

plt.bar(product, price, color=colors, edgecolor="black")
plt.title("Electronics-Prices")
plt.xlabel("products")
plt.ylabel("prices")
plt.show()
```

# 2. Scatter plot: 
### A scatter plot is visualization used to study the relationship between two numerical values.

## Scatter plot can help us identify:
1. Relationship - Whether two variables appear to be related.
2. Correlation - Whether variables tend to increase or decrease together.
3. Direction - Whether the relationship appears positive or negative.
4. Cluster - Groups of observation with similar characteristics.
5. Outliers - Observations that are far away from the general pattern.
6. Patterns - General structures or trends in the observations.

```python
# e.g.

hours_studied = [5, 4, 2, 1, 4, 5]
marks = [85, 70, 35, 20, 65, 40]

plt.scatter(hours_studied, marks)
plt.show()
```

```python
# X-label, Y-label, Color, Title, Grid, Edge-color:

hours_studied = [5, 4, 2, 1, 4, 5]
marks = [85, 70, 35, 20, 65, 40]

plt.scatter(hours_studied, marks, color="red", edgecolors="black")
plt.xlabel("hours")
plt.ylabel("marks")
plt.title("hours studied vs marks")
plt.grid()
plt.show()
```

# Subplots:
### A subplot allows us to display multiple plots in a single figure. This is useful when we want to compare different visualizations or datasets side-by-side.

```python
# e.g.

month = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [10, 15, 12, 20, 18, 20]
profit = [5, 8, 7, 12, 10, 10]

fig, ax = plt.subplots(1, 2) # We can also make it as (2, 1).

ax[0].plot(month, sales)
ax[0].set_title("sales")

ax[1].plot(month, profit)
ax[1].set_title("profit")
```

```python
# X-label, Y-label, Figure Title:

month = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [10, 15, 12, 20, 18, 20]
profit = [5, 8, 7, 12, 10, 10]

fig, ax = plt.subplots(1, 2)  # We can also make it as (2, 1).

ax[0].plot(month, sales)
ax[0].set_title("sales")
ax[0].set_xlabel("month")
ax[0].set_ylabel("sales")

ax[1].plot(month, profit)
ax[1].set_title("profit")
ax[1].set_xlabel("month") # We can also use share-x and share-y instead of x-label and y-label.
ax[1].set_ylabel("profit")

fig.suptitle("Performance")
plt.tight_layout() # To provide gap between two charts


# For downloading figure:

plt.savefig("Performance.png")
```

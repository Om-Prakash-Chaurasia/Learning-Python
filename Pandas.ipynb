## Series

```python
# 1. Series:

# Creating Series data with same data type. If we try to put different data type then it throws error as all the data type initially were of type int.
series1 = pd.Series([1, 2, 3, 4, 5])
print(series1)

# If the series have different data type together then we can update a value with any data type as it has object data type.
series2 = pd.Series([1, "Hello", "test", None, True, 99.99])
print(series2)
series2[2] = False
print(series2)

# If we create series with dictionary data type where these data can be accessed with keyName as keys act like an index here.
series3 = pd.Series({
    "name": "Om",
    "age": 26,
    "place": "Bangalore",
    "courses": ["Python", "Java", "JavaScript"],
    "tuple": (1, 2, 3)
})

print(series3)
print(series3["name"])
print(series3["courses"])
```

```python
# If we want to change the index type then we can do it as shown below but you will have to pass index value for each element.
series1.index = ["a", "b", "c", "d", "e"]
print(series1)

series1.index = [1, 3, 4, 2, 5]
print(series1)

# We can also sort the element on the basis of index or values.
series1 = series1.sort_index()
print(series1)

series1 = series1.sort_values()
print(series1)
```

```python
# Sorting different data type values
series4 = pd.Series([6, 3, None, True])
print(series4)
print(series4.sort_values())

# We can also pass a random index value but we can not use sort by index here whereas sorting by value is still possible.
series1.index = [1, 3, 2, 4, "e"]
print(series1)

# print(series1.sort_index()) # It throws TypeError: '<' not supported between instances of 'str' and 'int'

# Accessing individual element from the series.
print(series1["e"])
```

```python
# Updating value at individual index.
series1["e"] = 96
print(series1)

# Adding new index and its value.
series1["f"] = 987
print(series1)

# Deleting individual index and its value.
series1 = series1.drop("e")
print(series1)

series1 = series1.drop(2)
print(series1)

del series4[2]
print(series4)
```

```python
# We can assign index manually during series creation itself.
series5 = pd.Series([98, 95, 67, 90, None], index = ["r", "c", 2, 11, None])
print(series5)

# But if you notice the output of above code then you will fine None index has value of NaN this is because dtype of the series is float64 but if dtype changes then this NaN may also change to None. e.g. is shown below where dtype is object.
series6 = pd.Series([98, 95, 67, 90, None, True], index=["r", "c", 2, 11, None, True])
print(series6)

# Checking data type of series.
print(series6.dtype)

# Checking size of the series.
print(series6.size)

# Check the size of rows and columns.
print(series6.shape) # Here, the output is tuple and since we have only 1 column so we are getting no. of rows only as output.

# info(): Give details about the complete series.
print(series6.info())

```

```python
# describe(): Pandas creates a new series by calculating the summary statistics and returns this series. Therefore, the dtype of new series is shown as int64.
print(series6.describe()) 
```

## DataFrame

```python
# Think of a DataFrame as an Excel spreadsheet or SQL table.

# A DataFrame is a two-dimensional data structure consisting of:

# Rows
# Columns
# Indexes
# Values

# For example:

# index	Name	Age	Marks
# 0	    Rahul	20	85
# 1	    Priya	21	90
# 2	    Amit	19	78

# This can be represented in Pandas as:

import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Amit"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)

# Here in the output: Name, Age, and Marks are Columns and 0, 1, 2 are index values.
```

### Creating a DataFrame

```python
# Creating a DataFrame: There are several ways to create a DataFrame some of them are demonstrated below:

# 1. Using a Dictionary: This is the commonly used method.

data1 = {
    "Name": ["Neha", "Vidhya", "Sonam"],
    "Age": [25, 27, 26],
    "Marks": [87, 78, 77]
}

df1 = pd.DataFrame(data1)
print(df1)

# 2. Using a List of Lists:

data2 = [
    ["Lewis", 25, 87],
    ["Max", 27, 28],
    ["Lando", 26, 77]
]

df2 = pd.DataFrame(data2, columns=["Name", "Age", "Marks"]) # Make sure you pass column param other wise instead of columnName you will get index value there too.
print(df2)

# 3. Using a List of Dictionaries: This approach is particularly useful when working with JSON-like data.

data3 = [
    {"Name": "Doraemon", "Age": 56, "Marks": 98},
    {"Name": "Novita", "Age": 67, "Marks": 67},
    {"Name": "Suzuka", "Age": 65, "Marks": 88}
]

df3 = pd.DataFrame(data3)
print(df3)
```

### Accessing Data from DataFrame

```python
# Since, DataFrame has rows and columns we can get all columns but not rows as rows does not have names instead have index.

# 1. columns: Fetching all column names:
print("All column names:\n", df1.columns, sep="")

# 2. to_list(): We can convert output of above code as Python list by doing:
print("Column names as Python List:\n", df1.columns.to_list(), sep="")

# 3. len(): This will get the number of rows.
print("Length of rows:\n", len(df1), sep="")

# 4. shape: This returns the number of rows and columns in the DataFrame.
print("Shape:\n", df1.shape, sep="") # The output will have (rows, columns)
```

```python
# 5. Accessing Columns: We can access column using its name.
print("Accessing all values of Name column:\n", df2["Name"], sep="") # Notice that the result is a Series.
print("Converting Names into list:\n", df2["Name"].to_list(), sep="")

# Alternatively:
print(df2.Name)

# However, df["Name"] is generally preferred because it works with column names that contain spaces

# 6. Accessing multiple columns:
print(df2[["Name", "Marks"]]) # Here, we need to pass columnNames as List and we can not use dot operator like before.
```

### Indexing in DataFrame

```python
# Indexing in DataFrame: Indexing means selecting specific rows or columns. There are two important methods for this:

# 1. loc (Label-Based Indexing): loc is used when you want to access data using labels.
print(df3.loc[0]) # This gets the first row.

# a. Access a particular value.
print(df3.loc[2, "Name"])

# b. Accessing multiple rows using loc:
print(
    df3.loc[0:1]
)  # This will give all columns but only 2 rows. Here, ending index is inclusive.

# c. Accessing specific number of rows and columns:
print(
    df3.loc[0:1, "Name":"Age"]
)  # Here, we are specifying both rows and columns as [rowStart : rowEnd, columnStart : columnEnd]

print(df3.loc[:, "Name":"Age"])  # This will give all rows but only 2 columns.
```

```python
# 2. iloc (Position-Based Indexing): iloc means integer-location based indexing. It works using numerical positions.

print(df3.iloc[0]) # This gets the first row.

# a. Access a particular value.
print(df3.iloc[2, 0]) # Here, instead of passing columnName i.e. "Name" we are passing index value of Name column.

# b. Accessing multiple rows using iloc:
print(
    df3.iloc[0:1]
)  # This will give all columns but only 1 row because here, ending index is exclusive.

# c. Accessing specific number of rows and columns:
print(
    df3.iloc[0:2, 0:2]
)  # Here, we are specifying both rows and columns as [rowStart : rowEnd, columnStart : columnEnd]. Ending index are exclusive.

print(df3.iloc[:, 0:2])  # This will give all rows but only 2 columns because here ending index is exclusive.
```

### Conversion in Pandas

```python
# Conversion in Pandas: Pandas provides many methods for converting data. Some of them are demonstrated below:

# 1. Converting DataFrame into List:
print(df3["Name"].to_list())

# 2. Converting DataFrame into dictionary:
print(df3["Name"].to_dict())
print(df3.to_dict())

# 3. DataFrame to NumPy Array:
print(df3[["Name", "Age"]].to_numpy())
print(df3.to_numpy())

# 4. DataFrame to CSV:
df3.to_csv("students.csv", index=False) # Here, index is True by default, so, if you don't want index then pass index=False.

# 5. CSV to DataFrame:
df9 = pd.read_csv("students.csv")
print(df9)
```

### Filtering Data

```python
# Filtering Data: Filtering is one of the most useful Pandas features.

# Suppose we want students whose marks are greater than 80.
result = df1[df1["Marks"] > 80]
print(result)

res = df1[df1["Age"] >= 26]
print(res)

# Multiple Conditions: We can pass multiple conditions too.

# AND (&):
# Students who are at least 26 years old and have marks greater than 75:
res = df1[
    (df1["Age"] >= 26) & 
    (df1["Marks"] > 75)
    ] # Unlike Python here we have to use '&' for 'and' and '|' for 'or'. 
print(res)

# OR(|):
res = df1[
    (df1["Marks"] > 80) |
    (df1["Age"] <= 20)
]
print(res)
```

### Missing Data:

```python
# Missing Data: Real-world datasets often contain missing values. e.g. is shown below:

df = pd.DataFrame({
    "Name": ["Tony Stark", "Captain America", "Thor"],
    "Age": [45, 105, None],
    "Marks": [99, None, 45]
})

print(df)

# Here, in output NaN means a missing numerical value.
```

### To check missing values:

```python
# 1. isna(): It checks for missing values.
print(df.isna())

# 2. isna().sum(): Count missing values.
print(df.isna().sum())

# 3. dropna(): The rows containing missing values are removed.
# print(df.dropna())

# 4. fillna(): Instead of removing missing values, we can replace them.
print(df.fillna(20)) 
# Fill all missing values with 20, which is not correct because one missing value is age and another is marks so we need to fill data appropriately. i.e.

df["Age"] = df["Age"].fillna(78)
print(df)  # Now the missing age is replaced with 78.

# mean(): Generally we don't fill missing value with random value instead we replace it with mean. e.g. is shown below:
df["Marks"] = df["Marks"].fillna(df["Marks"].mean()) # This is how we commonly do in data processing.
print(df)
```

### Text Data

```python
# Text Data: Pandas provides the .str accessor for working with text. e.g. is shown below:

df = pd.DataFrame({
    "Name": ["black widow", "ant-man", "Scarlet WITCH"]
})

print(df)

# 1. str.upper(): Converts text to upper case.
df["Name"] = df["Name"].str.upper()
print(df)

# 2. str.lower(): Converts text to lower case.
df["Name"] = df["Name"].str.lower()
print(df)

# 3. str.len(): Finds string length.
df["Length"] = df["Name"].str.len()
print(df)

# 4. str.contains(): Check whether text contains something, it is case-sensitive. It gives boolean result.
result = df.Name.str.contains("Ant")
print(result)

# To make it case-insensitive:
result = df.Name.str.contains("Ant", case=False)
print(result)

# 5. str.replace(): If we want to replace some text data then we can use this.
df.Name = df.Name.str.replace("ant", "Spider", case=False)
print(df)
```

### Concatenating DataFrame

```python
# Concatenating DataFrame: Concatenation means putting DataFrames together. e.g.

# Suppose we have:

d1 = pd.DataFrame({
    "Name": ["Rahul", "Priya"],
    "Marks": [85, 90]
})

print("Table 2:\n",d1, sep="")

d2 = pd.DataFrame({
    "Name": ["Amit", "Sneha"],
    "Marks": [78, 88]
})

print("Table 1:\n",d2, sep="")

# 1. Vertical Concatenating: By default concatenation is vertically because axis is zero.
result = pd.concat([d1, d2])
print(result) 
# Here, if you see output the indices has not changed just two dataFrames got combined because by default ignore_index is False.

# If we want to reset indices then we have use ignore_index=True
result = pd.concat([d1, d2], ignore_index=True)
print(result)

# 2. Horizontal Concatenation: If we want to concat horizontally then we use axis=1 which means combine column-wise.
result = pd.concat([d1, d2], axis=1)
print(result)
```

### Merging DataFrame

```python
# Merging DataFrame: In Pandas, merge() is used to combine two DataFrames based on one or more common columns, similar to JOIN operations in SQL. e.g.

# Suppose we have students information:
students = pd.DataFrame({
    "studentId": [1, 2, 3, 4],
    "studentName": ["Amit", "Rahul", "Priya", "Sanjay"]
})

print("Students Table:\n", students, sep="")

# And marks:
marks = pd.DataFrame({
    "studentId": [2, 3, 4, 5],
    "marks": [88, 92, 95, 86]
})

print("Marks Table:\n", marks, sep="")

# We can merge these two tables using "studentId", as shown below:
result = pd.merge(
    students,
    marks,
    on="studentId",  # This is the important argument in merging data. It tells Pandas which column should be used to match the records.
)

print(result)
```

### Types of Merge:

```python
# Types of Merge: Pandas support several types of Merge. The main types of merge are:

# 1. Inner Merge
# 2. Left Merge
# 3. Right Merge
# 4. Outer Merge
# 5. Cross Merge

# Notice that:

# student_id 2, 3, 4 exists in both DataFrames.
# student_id 1 exists only in students.
# student_id 5 exists only in marks.

# This will help us understand the different merge types.

# 1. Inner Merge: An inner merge returns only the rows that have matching values in both DataFrames.

# Syntax: pd.merge(df1, df2, on="column_name", how="inner")

result = pd.merge( students, marks, on="studentId", how="inner")

print("Inner Merge:\n", result, sep="")
# Student 1 and student 5 are excluded as these are not common.

# 2. Left Merge: A left merge keeps all rows from the left DataFrame and adds matching information from the right DataFrame.

# Syntax: pd.merge(df1, df2, on="column_name", how="left")

result = pd.merge( students, marks, on="studentId", how="left")

print("Left Merge:\n", result, sep="")

# Notice student 1 is still present even though there is no corresponding record in marks. Because there are no marks available: marks = NaN
# Suppose during your semester exam in some cases if the student is absent then name of the student exists in the list but marks for that student does not exists.

# 3. Right Merge: A right merge keeps all rows from the right DataFrame and adds matching information from the left DataFrame.

# Syntax: pd.merge(df1, df2, on="column_name", how="right")

result = pd.merge( students, marks, on="studentId", how="right")

print("Right Merge:\n", result, sep="")
# Notice student 5 is present in marks, but there is no corresponding student in students.
# Suppose during your semester exam in some cases if the student writes the wrong Roll No. then marks of the student exists but we don't know whose marks it is.

# 4. Outer Merge: An outer merge keeps all rows from both DataFrames.

# Syntax: pd.merge(df1, df2, on="column_name", how="outer")

result = pd.merge(students, marks, on="studentId", how="outer")

print("Outer Merge:\n", result, sep="")

# Here:
# Student 1 comes from students.
# Students 2, 3, 4 exist in both.
# Student ID 5 comes from marks.
```

```python
# 5. Cross Merge: A cross merge creates the Cartesian product of the two DataFrames. In simple terms: Every row of DataFrame 1 is combined with every row of DataFrame 2. e.g.

df1 = pd.DataFrame({
    "Color": ["Pepsi", "Fanta"]
})

df2 = pd.DataFrame({
    "Size": ["Small", "Large"]
})

result = pd.merge(df1, df2, how="cross")

print("Cross Merge:\n", result, sep="")

# Why 4 rows?
# 2 colors × 2 sizes = 4 combinations
# For example, this can be useful when generating all possible combinations of products and sizes.
```

### Joining DataFrame

```python
# Joining DataFrames: join() is another way of combining DataFrames, commonly based on their indexes. This is very similar to Merge in Pandas. The main difference is that join() is generally designed to join using the index. e.g.

employees = pd.DataFrame({
    "name": ["Amit", "Riya", "Rahul", "Sneha"]
    }, index=[101, 102, 103, 104])

print("Employees Table:\n", employees, sep="")

salary = pd.DataFrame(
    {
        "salary": [50000, 60000, 55000, 70000]
    }, index=[102, 103, 104, 105]
)

print("Salary Table:\n", salary, sep="")

# 1. Inner Join:
result = employees.join(salary, how="inner")

print("Inner Join:\n", result, sep="")

result = pd.merge(df1, df2, on="studentId", how="inner")

# 2. Left Join:

result = employees.join(salary, how="left")

print("Left Join:\n", result, sep="")

# 3. Right Join:
result = employees.join(salary, how="right")

print("Right Join:\n", result, sep="")

# 4. Outer Join:
result = employees.join(salary, how="outer")

print("Outer Join:\n", result, sep="")

# 5. Cross Join:

result = employees.join(salary, how="cross")

print("Cross Join:\n", result, sep="")

# So you can remember:
# merge() → Match columns
# join()  → Match indexes

# Although join() can also be configured to use columns, and merge() can also work with indexes, so this is a rule of thumb rather than a strict limitation.
```

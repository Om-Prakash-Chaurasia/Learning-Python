```python
# Variable arguments :- O/P is tuple

def add(*nums):
    sum = 0
    for num in nums:
        sum += num 
    return sum

add(5, 10)
```

```python
# Keyword arguments :- O/P is dictionary

def student(**details):
    print(details)
student(name="Om", age=25, city="Mangalore")
```

```python
# Printing individual properties

def details(**values):
    print("Name:", values["name"])
    print("Age:", values["age"])
    print("Department:", values["dept"])
    print(values["address"])

details(name = "Om", age = 25, dept = "FSD", address = "123 ABC")
```

```python
# Looping on Keyword Argument:

def student(**details):
    for key, value in details.items():
        print(key, "=", value)

student(name = "Om", age = 25, city = "Mangalore", course = "Python")
```

```python
# Combining *agrs and **kwargs => Here *agrs should always come before **kwargs

def demo(*args, state, **kwargs):
    print(state)
    print("Arguments:", args)
    print("Keyword Arguments:", kwargs)


demo(10, 20, 30, state="Delhi", name="Om", age=25)
```

Lambda Function

Syntax:
lambda arguments: expression

```python
# def square(x):
#     print(x * x)
# square(6)

square = lambda x: x * x
print(square(5))
```

```python
# def add(a, b):
#     print(a + b)
# add(5, 7)

add = lambda a, b: a + b
print(add(6, 7))
```

```python
# Odd or even using lambda function

# def check(n):
#     if n%2 == 0:
#         print("Even")
#     else:
#         print("Odd")
# check(45)

check = lambda n: "Even" if n%2 == 0 else "Odd"
print(check(34))
```

```python
# Lambda with map()

nums = [1, 2, 3, 4, 5]

double = map(lambda n: n * 2, nums)
print(list(double))

square = map(lambda n: n * n, nums)
print(list(square))

cube = map(lambda n: n * n * n, nums)
print(list(cube))
```

```python
# Lambda with filter

nums = [34, 48, 12, 95, 77, 53, 64]

even = filter(lambda n: n%2 == 0, nums)
print(list(even))

odd = filter(lambda n: n%2 == 1, nums)
print(list(odd))
```
